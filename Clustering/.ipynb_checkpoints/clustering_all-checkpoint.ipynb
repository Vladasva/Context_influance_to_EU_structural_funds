```python
import pandas as pd
import numpy as np
from sklearn.cluster import KMeans 
from sklearn.preprocessing import StandardScaler
import seaborn as sns
from sklearn.metrics import silhouette_score
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
```

```python
dk = pd.read_csv("/Users/vladasverkelis/Documents/Doktorantūra/Straipsnis_1/Context_influance_to_EU_structural_funds/Data/Context/Denmark_2022.csv")

dk = dk.rename({"Unnamed: 0":"Context", "22":"Denmark"}, axis=1)

de = pd.read_csv("/Users/vladasverkelis/Documents/Doktorantūra/Straipsnis_1/Context_influance_to_EU_structural_funds/Data/Context/Germany_2022.csv")

de = de.rename({"Unnamed: 0":"Context", "22":"Germany"}, axis=1)

pl = pd.read_csv("/Users/vladasverkelis/Documents/Doktorantūra/Straipsnis_1/Context_influance_to_EU_structural_funds/Data/Context/Poland_2022.csv")

pl = pl.rename({"Unnamed: 0":"Context", "22":"Poland"}, axis=1)

lt = pd.read_csv("/Users/vladasverkelis/Documents/Doktorantūra/Straipsnis_1/Context_influance_to_EU_structural_funds/Data/Context/Lithuania_2022.csv")

lt = lt.rename({"Unnamed: 0":"Context", "22":"Lithuania"}, axis=1)

lv = pd.read_csv("/Users/vladasverkelis/Documents/Doktorantūra/Straipsnis_1/Context_influance_to_EU_structural_funds/Data/Context/Latvia_2022.csv")

lv = lv.rename({"Unnamed: 0":"Context", "22":"Latvia"}, axis=1)

ee = pd.read_csv("/Users/vladasverkelis/Documents/Doktorantūra/Straipsnis_1/Context_influance_to_EU_structural_funds/Data/Context/Estonia_2022.csv")

ee = ee.rename({"Unnamed: 0":"Context", "22":"Estonia"}, axis=1)

fi = pd.read_csv("/Users/vladasverkelis/Documents/Doktorantūra/Straipsnis_1/Context_influance_to_EU_structural_funds/Data/Context/Finland_2022.csv")

fi = fi.rename({"Unnamed: 0":"Context", "22":"Finland"}, axis=1)

se = pd.read_csv("/Users/vladasverkelis/Documents/Doktorantūra/Straipsnis_1/Context_influance_to_EU_structural_funds/Data/Context/Sweden_2022.csv")

se = se.rename({"Unnamed: 0":"Context", "22":"Sweden"}, axis=1)
```

```python
all_countries = dk
all_countries = pd.merge(all_countries, de, on='Context', how='outer')
all_countries = pd.merge(all_countries, pl, on='Context', how='outer')
all_countries = pd.merge(all_countries, lt, on='Context', how='outer')
all_countries = pd.merge(all_countries, lv, on='Context', how='outer')
all_countries = pd.merge(all_countries, ee, on='Context', how='outer')
all_countries = pd.merge(all_countries, fi, on='Context', how='outer')
all_countries = pd.merge(all_countries, se, on='Context', how='outer')
```

```python
all_countries
```

```python
all_countries = all_countries.transpose()
```

```python
all_countries.columns = all_countries.iloc[0]
```

```python
all_countries = all_countries[1:]
all_countries = all_countries.reset_index()
```

```python
all_countries = all_countries.rename({"index":"Country"}, axis =1)
```

```python
all_countries
```

```python
numeric_data = all_countries.drop(columns=['Country'], axis=1)
```

```python
labels = all_countries['Country']
```

```python
sc = StandardScaler()

X = sc.fit_transform(numeric_data)
```

```python
print(f"Number of samples: {len(all_countries)}")

K_range = range(2, 8)

inertias = []
silhouette_scores = []

for k in K_range:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(X)
    inertias.append(kmeans.inertia_)
    silhouette_scores.append(silhouette_score(X, kmeans.labels_))

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

ax1.plot(K_range, inertias, 'bo-')
ax1.set_xlabel('Number of clusters (k)')
ax1.set_ylabel('Inertia')
ax1.set_title('Elbow Method')
ax1.grid(True)

ax2.plot(K_range, silhouette_scores, 'ro-')
ax2.set_xlabel('Number of clusters (k)')
ax2.set_ylabel('Silhouette Score')
ax2.set_title('Silhouette Analysis')
ax2.grid(True)

plt.tight_layout()
plt.show()
```

```python
optimal_k = 3
kmeans = KMeans(n_clusters=optimal_k, random_state=42, n_init=10)
all_countries['Cluster'] = kmeans.fit_predict(X)

print(f"\n{'='*60}")
print(f"K-Means Clustering Results (k={optimal_k})")
print(f"{'='*60}\n")
```

```python
colors = ['red', 'blue', 'green']
cluster_names = ["Cluster 1", "Cluster 2", "Cluster 3"] 

pca = PCA(n_components=2, random_state=42)
X_pca = pca.fit_transform(X)

plt.figure(figsize=(14, 10)) 
for i in range(optimal_k):
    cluster_indices = all_countries['Cluster'] == i
    plt.scatter(X_pca[cluster_indices, 0],
                X_pca[cluster_indices, 1],
                c=colors[i],
                label=f'Cluster {i} ({cluster_names[i]})',
                s=200, 
                alpha=0.6,
                edgecolors='black',
                linewidth=1.5)
    

    cluster_data = all_countries[cluster_indices]
    for idx, row in cluster_data.iterrows():
        country_idx = all_countries.index.get_loc(idx)
        plt.annotate(row['Country'], 
                    (X_pca[country_idx, 0], X_pca[country_idx, 1]),
                    fontsize=10, 
                    fontweight='bold',
                    xytext=(5, 5), 
                    textcoords='offset points',
                    bbox=dict(boxstyle='round,pad=0.3', facecolor=colors[i], alpha=0.3))


centroids_pca = pca.transform(kmeans.cluster_centers_)
plt.scatter(centroids_pca[:, 0], centroids_pca[:, 1],
            c='black', marker='X', s=400, edgecolors='yellow',
            linewidths=3, label='Centroids', zorder=10)

plt.xlabel(f'PC1 ({pca.explained_variance_ratio_[0]:.2%} variance)', 
           fontsize=13, fontweight='bold')
plt.ylabel(f'PC2 ({pca.explained_variance_ratio_[1]:.2%} variance)', 
           fontsize=13, fontweight='bold')
plt.title(f'K-Means Clustering (k={optimal_k}) - PCA 2D Projection', 
          fontsize=15, fontweight='bold')
plt.legend(loc='best', fontsize=11)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()


print(f"\nTotal variance explained by 2 PCs: {pca.explained_variance_ratio_.sum():.2%}")
```

```python
optimal_k = 3

kmeans = KMeans(n_clusters=optimal_k, random_state=42, n_init=10)
cluster_labels = kmeans.fit_predict(X)

all_countries['Cluster'] = cluster_labels

print("="*60)
print("CLUSTERING STATISTICS")
print("="*60)
print(f"\nNumber of clusters: {optimal_k}")
print(f"Inertia: {kmeans.inertia_:.2f}")
print(f"Silhouette Score: {silhouette_score(X, cluster_labels):.3f}")

print("\n" + "-"*60)
print("Countries by Cluster:")
print("-"*60)
for cluster in range(optimal_k):
    countries = all_countries[all_countries['Cluster'] == cluster]['Country'].tolist()
    print(f"\nCluster {cluster} ({len(countries)} countries):")
    print(f"  {', '.join(countries)}")

print("\n" + "="*60)
print("CLUSTER CHARACTERISTICS (Mean Values)")
print("="*60)

feature_cols = [col for col in all_countries.columns if col not in ['Country', 'Cluster']]

cluster_means = all_countries.groupby('Cluster')[feature_cols].mean()
print("\n", cluster_means.round(2))

print("\n" + "="*60)
print("DEFINING VARIABLES FOR EACH CLUSTER")
print("="*60)

overall_mean = all_countries[feature_cols].mean()
overall_std = all_countries[feature_cols].std()

for cluster in range(optimal_k):
    print(f"\n{'='*60}")
    print(f"CLUSTER {cluster} - Distinctive Characteristics")
    print('='*60)
    
    cluster_data = all_countries[all_countries['Cluster'] == cluster][feature_cols]
    cluster_mean = cluster_data.mean()
    
    z_scores = (cluster_mean - overall_mean) / overall_std
    
    distinctive_features = z_scores.abs().sort_values(ascending=False)
    
    print("\nTop 5 Distinctive Variables:")
    for i, (feature, z_score) in enumerate(z_scores[distinctive_features.index[:10]].items(), 1):
        direction = "HIGH" if z_score > 0 else "LOW"
        print(f"{i}. {feature:10s}: {direction:4s} (z-score: {z_score:+.2f}, "
              f"mean: {cluster_mean[feature]:.2f} vs overall: {overall_mean[feature]:.2f})")

```
