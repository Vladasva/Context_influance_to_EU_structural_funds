```python
from scipy.stats import shapiro 
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
```

```python
df_dk = pd.read_csv("/Users/vladasverkelis/Documents/Doktorantūra/Straipsnis_1/Context_influance_to_EU_structural_funds/Data/y/dk_bn.csv")
```

```python
def shapiro_test(df):
    for feature in df.columns:
        if feature != "Unnamed: 0":
            print(feature)
            print(shapiro(df[feature]))
            
def fix_outliers_with_mean(df, column_name):
    Q1 = df[column_name].quantile(0.25)
    Q3 = df[column_name].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    clean_mean = df.loc[(df[column_name] >= lower) & (df[column_name] <= upper), column_name].mean()

    df.loc[(df[column_name] < lower) | (df[column_name] > upper), column_name] = clean_mean
    
    return df

def fix_outliers_with_mode(df, column_name):
    Q1 = df[column_name].quantile(0.25)
    Q3 = df[column_name].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    clean_mode = df.loc[(df[column_name] >= lower) & (df[column_name] <= upper), column_name].mode()[0]

    df.loc[(df[column_name] < lower) | (df[column_name] > upper), column_name] = clean_mode
    
    return df

def transform_log(df, feature):
    df_lg = df.copy() 
    df_lg[feature] = np.log1p(df_lg[feature])
    return df_lg

def transform_sqrt(df, feature):
    df_sq = df.copy()
    df_sq[feature] = np.sqrt(df_sq[feature])
    return df_sq

def transform_cbrt(df, feature):
    df_cb = df.copy()
    df_cb[feature] = np.cbrt(df_cb[feature])
    return df_cb
```

```python
shapiro_test(df_dk)
```

```python
df_dk = fix_outliers_with_mean(df_dk, "SSB")
```

```python
df_dk = fix_outliers_with_mean(df_dk, "PT")
```

```python
df_dk_lg = transform_log(df_dk, "SSB")
```

```python
df_dk_lg = transform_log(df_dk, "PT")
```

```python
df_dk_sq = transform_sqrt(df_dk, "SSB")
```

```python
df_dk_sq = transform_sqrt(df_dk, "PT")
```

```python
df_dk_cb = transform_cbrt(df_dk, "SSB")
```

```python
df_dk_cb = transform_cbrt(df_dk, "PT")
```

```python
shapiro_test(df_dk_sq)
```
